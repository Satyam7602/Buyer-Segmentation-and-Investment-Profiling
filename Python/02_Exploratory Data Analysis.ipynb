# Analysis 2: Gender Distribution

## Business Question

What is the gender distribution of buyers in the real estate market?

```python
df["gender"].value_counts()
```

```python
df["gender"].value_counts().plot(
    kind="bar",
    figsize=(7,5),
    title="Gender Distribution of Buyers"
)

plt.xlabel("Gender")
plt.ylabel("Number of Buyers")

plt.show()
```

```python
df["gender"].value_counts().plot(
    kind="pie",
    figsize=(6,6),
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Gender Distribution of Buyers")
plt.ylabel("")

plt.show()
```

## Key Findings

• A total of 7,305 buyers were analysed.

• Male buyers account for 3,723 customers (51.0%).

• Female buyers account for 3,582 customers (49.0%).

• The gender distribution is almost balanced, with only a slight dominance of male buyers.

## Business Insight

The gender distribution analysis provides valuable information about the composition of the customer base. If male buyers account for the majority of purchases, the company may consider maintaining marketing strategies that appeal to this segment while also creating campaigns to attract more female buyers. A balanced customer portfolio can improve market reach and support long-term business growth.

The gender distribution indicates that the real estate company attracts both male and female buyers almost equally. Male buyers represent a slightly larger share (51.0%) compared to female buyers (49.0%), suggesting that the company's products and marketing strategies appeal to a broad customer base. This balanced distribution reduces dependency on a single customer segment and provides opportunities to design inclusive marketing campaigns that target both genders effectively.

## Recommendation

• Continue maintaining gender-inclusive marketing campaigns.

• Develop personalized property recommendations based on customer preferences rather than gender alone.

• Conduct further analysis by combining gender with acquisition purpose, loan application status, and property type to identify deeper customer behaviour patterns.

• Monitor gender trends regularly to ensure balanced customer engagement across future marketing campaigns.

# Analysis 3: Country Distribution

## Business Question

Which countries have the highest number of property buyers?

```python
df["country"].value_counts()
```

```python
country_counts = df["country"].value_counts()

ax = country_counts.plot(
    kind="bar",
    figsize=(10,6),
    title="Number of Buyers by Country"
)

plt.xlabel("Country")
plt.ylabel("Number of Buyers")
plt.xticks(rotation=45)

for bar in ax.patches:
    ax.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        int(bar.get_height()),
        ha="center",
        va="bottom"
    )

plt.show()
```

## Key Findings

• A total of 7,305 buyers were analysed across 10 countries.

• The USA has the highest number of buyers (5,656), representing the largest customer base.

• The UK (348) and Canada (300) are the second and third largest markets.

• Denmark has the lowest number of buyers (52).

• Buyer distribution is highly concentrated in the USA, while the remaining countries contribute a much smaller share.

## Business Insight

The country-wise analysis indicates that the United States is the company's strongest market, accounting for the majority of property buyers. This suggests strong brand recognition and customer demand within the U.S. However, the relatively lower buyer counts in other countries indicate significant opportunities for international market expansion through localized marketing campaigns, strategic partnerships, and region-specific investment offerings.

## Recommendation

• Continue strengthening market leadership in the USA through customer retention and premium property offerings.

• Increase marketing investments in the UK and Canada to expand the customer base.

• Develop country-specific promotional campaigns for emerging markets such as Germany, France, Belgium, Australia, and Mexico.

• Conduct detailed market research to understand customer preferences in lower-performing countries before expanding operations.

# Analysis 7: Customer Satisfaction Score Analysis

## Business Question

How are customer satisfaction scores distributed, and what do they indicate about the overall customer experience?

```python
df["satisfaction_score"].value_counts().sort_index()
```

```python
score_counts = df["satisfaction_score"].value_counts().sort_index()

ax = score_counts.plot(
    kind="bar",
    figsize=(8,5),
    color="steelblue",
    title="Customer Satisfaction Score Distribution"
)

plt.xlabel("Satisfaction Score")
plt.ylabel("Number of Buyers")
plt.grid(axis="y", linestyle="--", alpha=0.4)

for bar in ax.patches:
    ax.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        int(bar.get_height()),
        ha="center",
        va="bottom",
        fontsize=10
    )

plt.show()
```

```python
df["satisfaction_score"].mean()
```

```python
df["satisfaction_score"].median()
```

## Key Findings

• A total of 7,305 customer satisfaction responses were analysed.

• The highest number of buyers gave a satisfaction score of 5 (1,531), followed closely by a score of 4 (1,518).

• The average customer satisfaction score is 3.05 out of 5.

• The median satisfaction score is 3, indicating that overall customer satisfaction is moderate.

• Although positive ratings (4 and 5) are slightly more frequent, a considerable number of customers also reported lower satisfaction scores (1 and 2).

## Business Insight

The customer satisfaction analysis indicates an overall average satisfaction score of 3.05 out of 5, suggesting that customers are moderately satisfied with the company's services. While many customers reported high satisfaction levels (scores of 4 and 5), a substantial proportion provided lower ratings, indicating opportunities to improve customer experience, service quality, and post-purchase support.

## Recommendation

• Conduct customer feedback surveys to identify the reasons behind lower satisfaction scores.

• Improve after-sales service and customer support.

• Enhance communication during the property purchasing process.

• Introduce customer loyalty and referral programs to improve long-term customer satisfaction.

• Monitor satisfaction scores regularly and establish service quality benchmarks.

# Analysis 9: Sale Price Analysis

## Business Question

What is the distribution of property sale prices, and what is the average transaction value?

```python
df["sale_price"] = (
    df["sale_price"]
    .replace("[$,]", "", regex=True)
    .astype(float)
)
```

```python
df["sale_price"].mean()
```

```python
print("Minimum Sale Price:", df["sale_price"].min())
print("Maximum Sale Price:", df["sale_price"].max())
```

```python
ax = df["sale_price"].plot(
    kind="hist",
    bins=20,
    figsize=(10,6),
    title="Distribution of Property Sale Prices"
)

plt.xlabel("Sale Price")
plt.ylabel("Number of Properties")
plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.show()
```

## Key Findings

• A total of 7,305 property transactions were analysed.

• The average property sale price is $345,072.00.

• The minimum recorded sale price is $97,402.80.

• The maximum recorded sale price is $736,652.27.

• Most property transactions are concentrated in the mid-price range, with comparatively fewer high-value luxury properties.

## Business Insight

The sale price analysis indicates that the company's portfolio is primarily positioned within the mid-priced residential market. The average transaction value of $345,072 suggests steady demand from middle-income buyers. While high-value luxury properties contribute to revenue, they represent a smaller share of total transactions. This balanced pricing structure allows the company to serve a broad customer base while maintaining opportunities for premium market growth.

## Recommendation

• Continue expanding the portfolio of mid-priced properties to meet the largest segment of customer demand.

• Develop premium marketing strategies for luxury properties to increase high-value sales.

• Introduce flexible financing options for higher-priced properties to improve affordability.

• Monitor pricing trends regularly to ensure competitive positioning within each market segment.
