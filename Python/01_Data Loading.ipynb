# Buyer Segmentation & Investment Profiling
## Data Loading and Data Preparation

```python
import pandas as pd
```

## Step 1: Load the Datasets

```python
clients = pd.read_excel("../Dataset/clients.xlsx")
properties = pd.read_excel("../Dataset/properties.xlsx")
```

## Step 2: Preview the Data

```python
clients.head()
```

```python
properties.head()
```

## Step 3: Check Dataset Size

```python
print("Clients Shape :", clients.shape)
print("Properties Shape :", properties.shape)
```

## Step 4: Check Data Types

```python
clients.info()
```

```python
properties.info()
```

## Step 5: Check Missing Values

```python
clients.isnull().sum()
```

```python
properties.isnull().sum()
```

## Step 6: Check Duplicate Records

```python
print("Client Duplicates :", clients.duplicated().sum())
print("Property Duplicates :", properties.duplicated().sum())
```

## Step 7: Merge the Datasets

```python
merged = pd.merge(
    clients,
    properties,
    left_on="client_id",
    right_on="client_ref",
    how="inner"
)
```

## Step 8: Check the Merged Dataset

```python
merged.head()
```

```python
merged.shape
```

## Step 9: Convert Sale Price to Numeric

```python
merged["sale_price"] = (
    merged["sale_price"]
        .astype(str)
        .str.replace("$", "", regex=False)
        .str.replace(",", "", regex=False)
)

merged["sale_price"] = pd.to_numeric(
    merged["sale_price"],
    errors="coerce"
)
```

## Step 10: Verify Data Type

```python
merged["sale_price"].dtype
```

```python
merged["sale_price"].head()
```

## Step 11: Save the Cleaned Dataset

```python
merged.to_excel("../Dataset/merged_data.xlsx", index=False)
```
